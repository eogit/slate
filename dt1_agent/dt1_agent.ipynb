# DT1 agent: hello world over MCP

A clean starting point for the agentic digital twin, independent of the peatland model.
The agent lives in `dt1_server.py`: an MCP server with a single tool, `hello`.
This notebook plays the role of the orchestrator. It launches the server as a subprocess,
talks to it over stdio with the MCP protocol, lists its tools and calls `hello`.

Requirements: `pip install -r requirements.txt` (the MCP Python SDK, v2).

```python
import sys
from pathlib import Path

from mcp import Client, StdioServerParameters

SERVER = Path("dt1_server.py").resolve()
params = StdioServerParameters(command=sys.executable, args=[str(SERVER)])
```

## Connect, discover the tools, and query the agent

```python
async with Client(params) as client:
    tools = await client.list_tools()
    print("Tools offered:", [t.name for t in tools.tools])

    result = await client.call_tool("hello")
    reply = result.content[0].text

print(reply)
```

```python
assert reply == "Hello from Claude's DT1 agent!"
```
